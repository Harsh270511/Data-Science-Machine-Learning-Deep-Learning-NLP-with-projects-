```python
# Instance Variable and Methods
class Dog:
  def __init__(self,name, age):
    self.name= name
    self.age= age

dog1= Dog("Bhondu", 3)
print(dog1)
print(f"Name of the Dog is: {dog1.name}")
print(dog1.age)
```

```python
dog2= Dog("dogie", 4)
print(dog2.name)
print(dog2.age)
```

```python
# Define a class with instance methods
class Dog:

  def __init__(self, name, age):
    self.name= name
    self.age= age

  def barK(self):
    print(f"{self.name} cry like hoowwoooo")

dog1= Dog("buddy", 4)
dog1.barK()
```

```python
#Modeling a bank account

#Define a class for bank account

class BankAccount:
  def __init__(self, owner, balance=0):
    self.owner= owner
    self.balance= balance

  def deposite(self, amount):
    self.balance +=amount
    print(f"{amount} is deposited. New balance is {self.balance}")

  def withdraw(self, amount):
    if amount > self.balance:
      print("Low balance")

    else:
      self.balance -=amount
      print(f"{amount} is withdraw. New balance is  {self.balance}")
  def get_balance(self):
    return self.balance

# Creating an account
customer1= BankAccount("Harsh", 1000)
customer1.get_balance()

```

```python
# withdraw money from bank account
#Calling instance methods
customer1= BankAccount("yash", 20)
customer1.withdraw(10)
```
