```python
#Inheritance (Single inheritance)
# Parent class

class Car:

  def __init__(self,doors, model, engineType ):
    self.doors= doors
    self.model= model
    self.engineType=  engineType

  def drive(self):
    print(f"The driver is going to drive {self.engineType} car")

  
```

```python
car1= Car(4,"bmw", "diesel")
car1.drive()
```

```python
#Inherite parent class in child class
 
class Tata(Car):
  def __init__(self, doors, model, engineType, is_selfDrive):
    super().__init__(doors, model, engineType)
    self.is_selfDrive= is_selfDrive

  def selfDrive(self):
    print(f"Tata supports self driving: {self.is_selfDrive}")
```

```python
tata1= Tata(4,"luxury", "electric", True)
tata1.selfDrive()
```

```python
tata1.drive()
```

```python
#Multiple Inheritance--> When a class inherits from more than one  base class

# Base class 1
class Animal:
  def __init__(self, name):
    self.name= name

#Base class 2
class tiger:
  def __init__(self, danger):
    self.danger= danger

#Derived class from muliple base class

class pet(Animal, tiger):
  def __init__(self, name, danger):
    Animal.__init__(self, name)
    tiger.__init__(self, danger)

  def speak(self):
    return f"{self.name} is danger ->{self.danger}"

animal1= pet("Lion", True)
print(f"Animal name is: {animal1.name}")
animal1.speak()
```
