```python
import seaborn as sns
```

```python
#Basic plotting using seaborn
tips = sns.load_dataset('tips')
tips
```

```python
# create a scatter plot
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title("Scatter plot of Total bill vs Tips")
plt.show()
```

```python
#Line plot
sns.lineplot(x='size', y='total_bill', data= tips, marker='o')
plt.title("Line plot of total Bill by size")
plt.show()
```

```python
#categorial bill
# Bar chart

sns.barplot(x='day', y='total_bill', data= tips)
plt.title("Bar chart of Day with Total bill")
```

```python
#Box plot
sns.boxplot(x='sex', y='total_bill', data= tips)
plt.title("Box plot of total_bill by sex")
plt.show()
```

```python
#Violin plot
sns.violinplot(x='sex', y='total_bill', data= tips)
plt.title("Violin plot of totel bill by sex")
plt.show()
```

```python
#Histrogram plot

sns.histplot(tips['total_bill'], bins=10, kde=True)
plt.title("Histogram plot of total bill")
plt.show()
```

```python
# KDE plot
sns.kdeplot(tips['total_bill'], fill=True)
```

```python
#Pair plot
sns.pairplot(tips)
```

```python
#Heat map
corr= tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True,cmap='coolwarm')
```

```python
import pandas as pd

sales_df= pd.read_csv('sales_data.csv')
sales_df.head()
```

```python
#Plot total sales by product
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category', y='Total Revenue', data=sales_df,estimator=sum)
plt.title("Bar plot of Total Revenue by Product Category")
plt.xlabel('Product')
plt.ylabel('Total sales')
plt.show()
```

```python
#Plot total sales by Region
plt.figure(figsize=(10,6))
sns.barplot(x='Region', y='Total Revenue', data=sales_df,estimator=sum)
plt.title("Bar plot of Total Revenue by  Region")
plt.xlabel('Region')
plt.ylabel('Total sales')
plt.show()
```
