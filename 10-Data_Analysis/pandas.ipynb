```python
# Series 
# Series with the help of list
data=[1,2,3,4]
series= pd.Series(data)
print("Series: \n",series)
print(type(series))
```

```python
# Series with the help of Dictionary
data={'a':1, 'b':2, 'c':3,'d':4}
d_series= pd.Series(data)
print("series: \n",d_series)
```

```python
data=[7,8,9]
idx=['i','ii','iii']
pd.Series(data, index=idx)
```

```python
#DataFrame
#Create a dataframe from dictionary of list
data={
  'Name':['Harsh', 'Yash', 'Ayush'],
  'Age':[20, 19, 17],
  'city':['Florida', 'New york', 'Bangalore']
}
idx=['a','b','c']
df= pd.DataFrame(data, index=idx)
print(df)
print(type(df))
```
