```python
#File operation- Read and Write files

#Read
with open('example.txt','r') as file:
  content= file.read()
  print(content)
```

```python
## read the file line by line

with open('example.txt','r') as file:
  for line in file:
    print(line.strip()) # .strip() removes new line character
```

```python
#write a file(overwriting)  

with open('example.txt', 'w') as file:
  file.write("Heii this is yash maurya who overwrite the content\n")
  file.write("I hope you will remember me Thankyou boysss\n")
```

```python
# write a file( without overwrite)
with open('example.txt', 'a') as file:
  file.write("\nThis operation is append operation means with overwriting" )
```

```python
#Writing a list of lines to a file
lines=['\nfirst line\n', 'Second line\n','third line\n']

with open('example.txt','a') as file:
  file.writelines(lines)
```

```python
#binary files

##writing to a binary files

data=b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
  file.write(data)
```

```python
# Reading a  file

with open('example.bin','rb') as file:
  content= file.read()
  print(content)
```

```python
# Read the content from source text file and write to the destination text file
#Copying a text file

with open('example.txt', 'r') as source_file:
  content = source_file.read()

with open('destination.txt', 'w') as destination_file:
  destination_file.write(content) 

```

```python
#Read the text file and count the number of lines, words, and character
def count_text_file(file_path):
  with open(file_path, 'r') as file: 
    lines=file.readlines()
    line_count= len(lines)
    word_count= sum(len(line.split()) for line in lines)
    char_count= sum(len(line) for line in lines)
  return line_count, word_count, char_count
file_path= 'example.txt'
lines, words, character= count_text_file(file_path)
print(f"Lines: {lines}, Words: {words}, Character: {character}")
```
