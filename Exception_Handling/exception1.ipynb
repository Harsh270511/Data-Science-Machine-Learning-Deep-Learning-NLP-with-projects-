```python
try:
  num = int(input("Enter the number"))
  res= 10/num
  print(res)
except ValueError:
  print("This is not valid input")
except ZeroDivisionError:
  print("Enter demoninator greater than zero")
except Exception as ex:
  print(ex)
```

```python
## try--> except--> else
try:
  num = int(input("Enter the number"))
  res= 12/num
except Exception as ex:
  print(ex)
else:
  print(f"The result is {res}")
```

```python
## try--> except--> else--> finally
try:
  num = int(input("Enter the number"))
  res= 12/num
except Exception as ex:
  print(ex)
else:
  print(f"The result is {res}")
finally:
  print("The execution is completed")
```
