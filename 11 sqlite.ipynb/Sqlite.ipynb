## SqLite


```python
import sqlite3
```

```python
#Connect to an SQLite Database
connection = sqlite3.connect('example.db')
connection
```

```python
#cursor will iterate all over rows and columns
cursor= connection.cursor()
```

```python
# Create a Table

cursor.execute('''
CREATE TABLE IF NOT EXISTS employee (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT
)
''')

# Commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employee
''')
```

```python
#Inserting the data in sqlite table

cursor.execute('''
Insert Into employee(name, age, department)
            values("Harsh", 20, "Data Scientist")
''')

cursor.execute('''
Insert Into employee(name, age, department)
            values("Yash", 19, "AI Engineer")
''')

cursor.execute('''
Insert Into employee(name, age, department)
            values("Akansha", 22, "CS")
''')

connection.commit()
```

```python
# Query the data from the table
cursor.execute('Select * from employee')

rows= cursor.fetchall()

#Read the data from the table
for row in rows:
  print(row)
```

```python
# Update the data in the table
cursor.execute('''  
UPDATE employee
Set age=200
Where name = "Yash"

''')

connection.commit()
# Query the data from the table
cursor.execute('Select * from employee')

rows= cursor.fetchall()

#Print the data from the table
for row in rows:
  print(row)
```

```python
#Delete the Data from the table
cursor.execute('''
Delete from employee
        Where name='Harsh'
''')
connection.commit()

# Query the data from the table
cursor.execute('Select * from employee')

rows= cursor.fetchall()

#Print the data from the table
for row in rows:
  print(row)
```

```python


# 1. Connect to database
connection = sqlite3.connect('student_data.db')

# 2. Create cursor
cursor = connection.cursor()

# 3. Create table
cursor.execute('''
CREATE TABLE IF NOT EXISTS students(
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    course TEXT,
    marks INTEGER
)
''')

# 4. Data to insert
student_data = [
    ('Harsh', 21, 'CSE', 85),
    ('Rahul', 20, 'IT', 78),
    ('Priya', 21, 'CSE', 92),
    ('Aman', 22, 'ECE', 74)
]

# 5. Insert multiple records
cursor.executemany('''
INSERT INTO students(name, age, course, marks)
VALUES(?,?,?,?)
''', student_data)

# 6. Save changes
connection.commit()

# 7. Retrieve data
cursor.execute("SELECT * FROM students")

data = cursor.fetchall()

# 8. Display data
print("Student Data:")
for row in data:
    print(row)

# 9. Close connection
connection.close()
```
