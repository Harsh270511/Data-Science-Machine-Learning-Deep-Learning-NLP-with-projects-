## Generator

```python
## In this case i start from 0 and function return only 0 and stop execution of function
def square(n):
  for i in range(3):
    return i**2

square(3)
  
```

### yield is useful here because, unlike return, it doesn't stop the function permanently. It pauses the function and resumes it on the next iteration.


```python
def sqr(n):
  for i in range(3):
    yield i**2

for j in sqr(3):
  print(j)
```

```python
# another way to get square of element using next() keyword
a= sqr(3)
a
```

```python
next(a)
```

```python
## Example2 
def my_generator():
  yield 1
  yield 2
  yield 3
```

```python
#accessing ex-2 values
gen = my_generator()
gen
```

```python
next(gen)
```

```python
#accessing ex-2 values
for val in gen:
  print(val)
```

```python
## Practical Example: Reading large file

def reading_large_file(file_path):
  with open(file_path, 'r') as file:
    for line in file:
      yield line
```

```python
file_path = 'large_file.txt'

for line in reading_large_file(file_path):
  print(line.strip())
```
