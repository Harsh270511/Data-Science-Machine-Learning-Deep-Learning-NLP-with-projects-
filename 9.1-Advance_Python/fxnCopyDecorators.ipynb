```python
##closure

def main_greet(msg):
  def sub_greet():
    print("Welcome to sub greet method")
    print(msg)
    print("I am inside sub_greet method")
  return sub_greet()
```

```python
main_greet("This is harsh greeting you all")
```

```python
##closure

def main_greet(fxn):
  def sub_greet():
    print("Welcome to sub greet method")
    fxn("harsh Maurya")
    print("I am inside sub_greet method")
  return sub_greet()
```

```python
main_greet(print)
```

```python
## Decorator
def main_welcome(func):

  def sub_welcome():
    print("line 1 welcome")
    func()
    print("line 2 welcome")
  sub_welcome()
```

```python
def course_intro():
  print("This is advance python course")

```

```python
main_welcome(course_intro)
```

```python
# Decorator syntax--> @main_welcome
@main_welcome
def course_intro():
  print("This is advance python course")
```

```python
# decorator example
def my_decorator(func):
  def wrapper():
    print("Something happend before")
    func()
    print("Something happend after")
  wrapper()
```

```python
@my_decorator

def say_hello():
  print("Hello !")
```

```python
# Decorator with argument

def repeat(n):
  def decorator(func):
    def wrapper(*args, **kwargs):
      for _ in range(n):
        func(*args, **kwargs)
    return wrapper
  return decorator
```

```python
@repeat(3)

def greet_you():
  print("Hello!")
```

```python
greet_you()
```
