### Function in details

```python
#Function with single parameter
def even_odd(num):
  """This is the program to check eith
  er number is even or odd"""
  if num %2==0:
    print("Even number")
  else:
    print("odd number")
"""Calling the function"""
even_odd(11)
```

```python
#funtion with mulitiple parameter
def add(a, b):
  return a+b

#calling function
print(add(12,12))
```

```python
#Default parameter
def greet(name="Guest"):
  print(f"Hii {name} Welcome to paradise")

#calling function without parameter
greet()
#calling function with parameter
greet("Harsh")
```

```python
# Variable length argument
#Positional and Keyword argument

###  Positional Argument
def print_num(*args):
  for num in args:
    print(num)
print_num(1,12,3,34,13,"harsh")
```

```python
###  Keyword argument

def print_details(**kwargs):
  for key, value in kwargs.items():
    print(f"{key} : {value}")

print_details(name="Harsh", age=20, country="India")
```

```python
## Combination of positional and keyword argument
def combination(*args, **kwargs):
    for value in args:
        print(f"Positional argument -> {value}")

    for key, value in kwargs.items():
        print(f"{key}: {value}")

combination(1, 2, 3, "harsh", "yash", name="Harsh", age=20, country="India")
```

```python
#return statement
def div(a,b):
  return a/b

div(16,23)
```

```python
# multiple parameter return statement
def div(a,b):
  return a,b,a/b

div(12,3)
```

```python
#Temperature conversion

def convert_temp(temp, unit):
  """This function converts temperature b/w Celsius and Fahrenheit"""
  if unit=='C':
    return temp*9/5 + 32 ## celsius to fahrenheit
  elif unit=='F':
    return (temp-32) * 5/9 ## fahrenheit to celsius
  else:
    return None

print(convert_temp(11, 'B'))
print(convert_temp(140, 'C'))
print(convert_temp(60, 'F'))
```
