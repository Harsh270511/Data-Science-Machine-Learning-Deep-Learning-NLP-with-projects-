```python
#Leap year 

year= int(input("Enter the year: "))

if year%4==0:
  if year %100==0:
    if year %400==0:
      print(year ,"is leap year")
    else:
      print(year, " is not leap year")
  else:
    print(year, " is leap year")
else:
  print(year, " is not leap year")
```

```python
# loop control statement

#break statement

for i in range(10):
  if(i==4):
    break
  print(i)
```

```python
#continue statement

for i in range(7):
  if( i==5):
    continue
  print(i )
```

```python
#pass statement

for i in range(5):
  if i==3:
    print("passing the number ", i)
    pass
  print(i)
```

```python
#nested for loop

for i in range(3):
  for j in range(2):
    print(f"i {i} and  j  {j}")
```
