```python
#Issue while assigning the value and hence we get concept of Shallow copy
student={'name': 'Yash Maurya', 'age': 20, 'height': 5.5, 12: 'experience'}
student_copy=student
print(student)
print(student_copy)

print()
student["name"]="The nun"
print(student)
print(student_copy)
#The id is same means both refer same memory location
print(id(student_copy))
print(id(student))
```

###A dictionary object is created in memory.

Memory
        +------------------------------+
student ---> {'name':'Yash Maurya',
              'age':20,
              'height':5.5,
              12:'experience'}
              
Step 2: Assignment
student_copy = student

This does not create a new dictionary.

Instead, both variables point to the same dictionary.

                +------------------------------+
student ------> | {'name':'Yash Maurya', ... } |
                +------------------------------+
student_copy ---^

There is only one dictionary in memory.

Step 3: Print
print(student)
print(student_copy)

Output

{'name': 'Yash Maurya', 'age': 20, 'height': 5.5, 12: 'experience'}
{'name': 'Yash Maurya', 'age': 20, 'height': 5.5, 12: 'experience'}

Both variables show the same dictionary.

Step 4: Modify dictionary
student["name"] = "The nun"

Since student and student_copy point to the same object, the dictionary itself changes.

                +------------------------------+
student ------> | {'name':'The nun', ... }     |
                +------------------------------+
student_copy ---^
Step 5: Print again
print(student)
print(student_copy)

Output

{'name': 'The nun', 'age': 20, 'height': 5.5, 12: 'experience'}
{'name': 'The nun', 'age': 20, 'height': 5.5, 12: 'experience'}

Both changed because they refer to the same dictionary.

Is this a shallow copy?

No.

This is simply reference assignment (aliasing).

student_copy = student

Both variables share the same object.

You can verify this:

print(id(student))
print(id(student_copy))

Output (example)

140291840243200
140291840243200

Same memory address ⇒ Same object.

```python
#Shallow copy
student={'name': 'Yash Maurya', 'age': 20, 'height': 5.5, 12: 'experience'}
student1_copy= student.copy()
print(student)
print(student1_copy)

print()

#updating only student
student["name"]="harsh bhaii"
print(student)
print(student1_copy)

print()
#The id is now differ means both refer different memory location
print(id(student1_copy))
print(id(student))
```

```python
#Iterating over Dictionaries

## You can use loops to iterate over dict, keys, values, and items

student={'name': 'Yash Maurya', 'age': 20, 'height': 5.5, 12: 'experience'}

## Iterating over keys
for key in student.keys():
  print(key)

print()
#Iterating over values
for value in student.values():
  print(value)

print()
#Iterating over items
for key, val in student.items():
  print(f"{key}: {val}")
```

```python
#Nested Dictionary
students={
  "student1": {"name": "harsh", "age": 20},
  "student2": {"name": "yash", "age":19}
}
print(students)
```

```python
#Acessing nested dictionary elements
print(students['student1']['name'])

print(students['student2']['name'])
```

```python
### Iterating over nested dict

for std_id, std_info in students.items():
  print(f"{std_id} : {std_info}")
  for key, val in std_info.items():
    print(f"{key}: {val}")
```

```python
## Dictionary Comprehension
sqr= {x: x**2 for x in range(5)}
print(sqr)
```

```python
### Conditional Dictionary Comprehension
even= {x: x**2 for x in range(5) if x%2==0}
print(even)
```

```python
#use a dict to count the frequency of the elements
num=[1,2,3,4,4,2,1,12,1,12,4,3,4,1,2]
fre={}

for number in num:
  if number in fre:
    fre[number]+=1
  else:
    fre[number]=1
print(fre)
```
