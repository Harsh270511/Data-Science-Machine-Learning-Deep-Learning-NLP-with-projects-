```python
#Creating a Tuple 

name=()
print(name)
print(type(name))
```

```python
# different way to create tuple 
lst= list()
print(type(lst))

tfl= tuple()
print(type(tfl))
```

```python
# conversion of list to tuple and vice versa
num= tuple([1,3,5,6,7])
print(num)

print("Conversion of list to tuple")
num1=[1,2,3,4,0]
tup = tuple(num1)
print(tup)
print(type(tup))
print(type(num1))

print("Conversion of tuple to list")

num2=(1,2,3,56)
lst= list(num2)
print(lst)
print(type(lst))
print(type(num2))
```

```python
#we can add different data type in tuple
tup= ("harsh", 89, 66.7, False)
print(tup)
```

```python
# Prove using id() function that list and tuple are diff. object in memory
lst=[1,23,4]
tup=(0,56,4443)

print(id(lst))
print(id(tup))
```

```python
#Accessing the element in tuple 
num=(12,"harsh","12.12", 12.1221, False)
print(type(num[3]))
print(num[0])
#slicing in tuple

print(num[1: 9])
print(num[::-1])
```

```python
#Opertions on tuple

#Concatination

num=(1,2,3,4)
val=('a','b','c')

concat_tuple= num + val
print(concat_tuple)

# Repetition (*)
result = num *3
print(result)
```

```python
#tuple immutable nature
num=(12,[11],14)
num[1].append("harsh")
print(num)
```

```python
#tuple methods--> index and count
num=(1,2,3,4,5,6)
print(num.index(6))
print(num.count(10))
```

```python
#tuple methods
num=("harsh", "maurya", "yash", "maurya")
print(num.count("yash"))
print(num.index("maurya"))
```

```python
#Packing and unpacking

##Packing
info="harsh", 21, True

print(info)
print(type(info))
```

```python
#Unpacking 
num=[10,20,30]
b,a,c=num
print(a)
print(b)
print(c)
print(type(a))
```

```python
#unpacking with * (Extended Operators)
nums=(1,2,3,4,5,6,9,8)

first, *middle, last= nums
print(first)
print(middle)
print(last)
```

```python
## unpacking with * (Extended Operators)
name=["harsh", "yash", 21, 20]
one,two, *three= name
print(one)
print(two)
print(three)
```

```python
# Nested Tuple
tuf= ((1,2,3,4,), ("harsh", "yash", "raju", "abc","pqr"))
print(tuf[0])
print(tuf[1][:2])
```

```python
### Inside list tuple 
lst= [[1,2,3,4], ("heii", "hello", "bonjour","nihow")]
print(lst[1])
print(lst[1][2])
```

```python
### Iteration over nested tuple
tup= ((1,2,3), ('a','b','c'), ("heii", "hello", "bonjour","nihow"))

for sub_tuple in tup:
  for items in sub_tuple:
    print(items, end=" ")
  print()
```
