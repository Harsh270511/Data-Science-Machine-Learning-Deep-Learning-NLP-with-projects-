#### set are unordered and don't allow duplicated


```python
# creating set

create_set= {1,2,3,4,5}
print(create_set)
print(type(create_set))
```

```python
#testing random things by own (we can define tuple,list,etc inside set)
lst=[1,2,3,4]
tup= (9,8,7,6)
my_set_list= set(lst)
my_set_tuple= set(tup)

print(my_set_list)
print(type(my_set_list))
print(my_set_tuple)
print(type(my_set_tuple))
```

```python
# other way to create set

new_set= set()
print(type(new_set))
```

```python
#defining set inside set doing by own
num1={1,2,3,4,5,6,7}
num2= set(num1)

print(num2)
```

```python
# no duplicates are printed and no order maintain
name={"harsh", "harsh", 3,4,4,3,2,8,9}
print(name)
```

```python
#Basic operation of set
#Adding 
my_set={1,2,5,6}
print(my_set)
my_set.add(100)
print(my_set)
```

```python
#Removing
my_set.remove(5)
print(my_set)
```

### If element is not present in the set but still we try to remove it, will cause an error "KeyError"

```python
my_set.remove(1000)
print(my_set)
```

```python
#remove an element safely
my_set.discard(12121)
print(my_set)
```

```python
#pop() method
num={1,2,3,4,5,6}
element= num.pop()
print(element)
print(num)
```

```python
#few more example of pop()
fruits={"apple","banana","mango"}
item= fruits.pop()
print(item)
print(fruits)
```

```python
#few more example of pop()
numss={121,343,456,3458}
itm= numss.pop()
print(itm)
```

```python
#add multiple elements(update())
val={10,20,30,40}
val.update([111,1111,11111,0])
print(val)
```

```python
#remove all elements(clear())
val.clear()
print(val)
```

```python
#set membership test
val={1,2,34}
print(34 in val)
print(100 in val)

print(1 not in val)
print(121 not in val)
```

```python
#Mathmatical opertions
set1= {1,2,4,6}
set2={1,4,7,9,10}

#union
union_res=set1.union(set2)
#two ways to union
print(union_res)
print(set1 | set2)

#intersection
int_res=set1.intersection(set2)
#Two ways for intersection
print(int_res)
print(set1 & set2)
```

```python
#intersection_update()
A={1,2,3,4}
B={3,4,5,6}

A.intersection_update(B)
print(A)
print(B)
```

```python
#difference()
A={1,2,3,4}
B={3,4,5,6}

#Two ways to difference
print(A-B)
print(A.difference(B))
```

```python
#difference_update()
A={1,2,3,4}
B={1,2,3,4}
res= A.difference_update(B)
print(A)
print(B)
```

```python
#symmetric_difference()
e=A.symmetric_difference(B)
print(e)


#symmetric_difference_update()
k=A.symmetric_difference_update(B)
print(k)
print(A)
print(B)

```

```python
set1={1,2,3}
set2={1,2,3,5}

#issubset()
print(set1.issubset(set2))
print(set1 <= set2)

#issuperset()
print(set1.issuperset(set2))
print(set1 >=set2)

#isdisjoint()
print(set1.isdisjoint(set2))
```

```python
#removing the duplicate from list
lst=[1,2,2,3,3,4,5]

set(lst)
```

```python
#counting unique word in string

txt="I am currently learning AI and ml by krish sir"
word=txt.split()

#convert list into set to get unique word
unq_word= set(word)
print(unq_word)
print(len(unq_word))
```
